```python

import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile

plt.rcParams["figure.figsize"] = (9, 3.5)
```

```python

# Create the index
n = np.arange(-5, 11)

# Create four signals
delta = (n == 0).astype(float)

step = (n >= 0).astype(float)

ramp = n * step

expo = 0.8**n * step

# Create graphs
fig, axes = plt.subplots(1, 4, figsize=(13, 3))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["Impulse", "Unit Step", "Ramp", "Exponential"]
):
    ax.stem(n, x)
    ax.set_title(title)
    ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

```python

n = np.arange(-5, 11)

a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(12, 3))

for ax, a in zip(axes, a_values):
    x = (a**n) * (n >= 0)
    ax.stem(n, x)
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

```python

n = np.arange(0, 41)

x1 = np.cos(2*np.pi*n/8)
x2 = np.cos(2*np.pi*n/7.5)

plt.plot(n, x1, label="cos(2πn/8)")
plt.plot(n, x2, label="cos(2πn/7.5)")

plt.xlabel("n")
plt.ylabel("Amplitude")
plt.title("Comparison of Cosine Signals")
plt.legend()
plt.grid()
plt.show()
```

```python

import os

fname = "my_voice.wav"

if os.path.exists(fname):

    fs, x = wavfile.read(fname)

    if x.ndim > 1:
        x = x.mean(axis=1)

    x = x / np.max(np.abs(x))

else:

    fs = 16000
    t = np.arange(0, 2, 1/fs)

    x = 0.5*np.sin(2*np.pi*440*t) * np.exp(-t)

t = np.arange(len(x)) / fs

print("Sampling rate:", fs, "Hz")
print("Duration:", len(x)/fs, "seconds")
print("Samples:", len(x))

plt.plot(t, x)
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("My Recording")
plt.show()
```

### Question 2

Why is a stem plot more appropriate than a continuous line for x(n)?

**Answer:**

A stem plot is more appropriate because a discrete-time signal is defined only at separate integer values of n. It shows each individual sample clearly. A continuous line may give the impression that the signal exists at every point in time.

It would violate the BIBO stability property (Bounded-Input Bounded-Output stability). This is because the impulse response grows without limit, meaning it is not absolutely summable. Therefore, the system would not be BIBO stable.
